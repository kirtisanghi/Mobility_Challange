# Object Detection Inference Guide (local)

## Python Environment Setup

This python notebook is a systematic guide to setup the python virtual environment which should be used to run the inference

The very first step to get the environment up and running is to create the `python virtualenv`, which can be done in 2 ways

1. Creation of python virtual env using `conda`
2. Creation of python virtual env using `venv`

#### Python env creation with conda

1. Install `miniconda`, if not already installed
    
    ```shell
    mkdir -p ~/miniconda3
    wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O ~/miniconda3/miniconda.sh
    bash ~/miniconda3/miniconda.sh -b -u -p ~/miniconda3
    rm -rf ~/miniconda3/miniconda.sh
    ```
    
    ```shell
    ~/miniconda3/bin/conda init bash
    ```

2. Create virtualenv using conda

    ```shell
    conda env create --file python-env.yml
    ```

3. Activate the virtualenv

    ```shell
    conda activate mobility-challenge-custom-env
    ```

#### Python env creation with venv

1. Create python virtualenv

    ```shell
    python -m venv mobility-challenge-custom-env
    ```

2. Activate the virtualenv
    
    ```shell
    source mobility-challenge-custom-env/bin/activate
    ```

3. Install the required dependencies

    ```shell
    pip install -r requirements.txt
    ```

## Ultralytics Package Built and Installation

Since we are using a fine tuned YOLOv8 model with custom object classes, we are cloning the `ultralytics` repo and adding the extra head and weights to the current architecture of yolo model and eventually concat the new head to the existing head. Hence, `ultralytics` package must be manually patched, built and installed in the python environment by executing following steps

1. Clone the `ultralytics` git repository in any temp location (outside of the `Mobility_Challenge` folder)

    ```shell
    git clone https://github.com/ultralytics/ultralytics
    ```

2. Navigate to `ultralytics` folder

    ```shell
    cd ultralytics
    ```

3. Git reset to commit `2071776a3672eb835d7c56cfff22114707765ac`

    ```shell
    git reset --hard 2071776a3672eb835d7c56cfff22114707765ac
    ```

4. Download and apply the patch

    ```shell
    wget https://gist.githubusercontent.com/Y-T-G/8f4fc0b78a0a559a06fe84ae4f359e6e/raw/17b1407fefeac86d089c4cf14f174c8bb44948af/add_head.patch
    git apply add_head.patch
    ```

5. Manually built and install the `ultralytics` package

    ```shell
    pip install -e .
    ```
